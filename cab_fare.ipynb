```python
rides = []

def calculate_fare(distance):
    if distance <= 5:
        return 50 + distance * 15
    elif distance <= 15:
        return 50 + distance * 12
    else:
        return 50 + distance * 10


def book_ride():
    name = input("Customer name: ")
    pickup = input("Pickup: ")
    destination = input("Destination: ")

    try:
        distance = float(input("Distance in KM: "))
        if distance <= 0 or not name.strip() or not pickup.strip() or not destination.strip():
            print("Enter valid details!")
            return
    except ValueError:
        print("Invalid distance!")
        return

    fare = calculate_fare(distance)

    rides.append({
        "name": name,
        "pickup": pickup,
        "destination": destination,
        "distance": distance,
        "fare": fare
    })

    print("Ride booked! Fare: ₹", fare)


def view_rides():
    for r in rides:
        print(r)

    if not rides:
        print("No rides found!")


def search_ride():
    name = input("Search customer: ")
    for r in rides:
        if r["name"].lower() == name.lower():
            print(r)
            return
    print("Ride not found!")


def revenue_report():
    total = sum(r["fare"] for r in rides)
    count = len(rides)
    print("Total rides:", count)
    print("Total revenue: ₹", total)
    print("Average fare:", total / count if count else 0)


while True:
    print("\n1.Book Ride  2.View Rides  3.Search Ride")
    print("4.Revenue Report  5.Exit")

    choice = input("Choose: ")

    if choice == "1":
        book_ride()
    elif choice == "2":
        view_rides()
    elif choice == "3":
        search_ride()
    elif choice == "4":
        revenue_report()
    elif choice == "5":
        print("Thank you!")
        break
    else:
        print("Invalid choice!")
```
